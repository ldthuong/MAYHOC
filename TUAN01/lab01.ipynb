# Bài 4: Slicing và Indexing

```python
a= torch.rand(3,4,5)
print("original tensor: ")
print(a)

print("\ntensor a[1]: ")
print(a[1])

print("\ntensor a[1:, 2:4] :")
print(a[1:, 2:4])
```

```python
print("lay 2 phan tu cuoi tren chieu thu 3: ")
print(a[: , : , -2])
```

## Nhận xét
- a có shape là torch.size([3,4,5])
- a[1] có shape là torch.size([4,5])
- a[1:, 2:4] có shape là torch.size([2,2,5])
- a[:, :, -2] là lấy 2 phần tử cuối trên chiều thứ 3

# Bài 5: Reshape và Flatten

```python
a= torch.rand(3,4)
print("original: ")
print(a)

print("\nFlatten: ")
print(a.ravel())

print("\nReshape to 3x2x2: ")
print(a.reshape(3,2,2))

print("\nReshape thanh kich thuoc 2x6: ")
print(a.reshape(2,6))

print("\nSo phan tu ban dau: ", a.numel())
print("\nSo phan tu sau khi reshape 3x2x2: ", a.reshape(3,2,2).numel())
print("\nSo phan tu sau khi reshape 2x6: ", a.reshape(2,6).numel())
```

## Nhận xét:
- tensor ban đầu 3x4 là 12
- sau khi reshape 3x2x2 là 12
- sau khi reshape 2x6 là 12
- vậy reshape chỉ thay đổi hình dạng của tensor chứ không thay đổi tổng số phần tử

# Bài 6: Phép toán và nhân ma trận

```python
a= torch.randn(2,3)
b= torch.randn(2,3)
print("a: ")
print(a)
print("\nb: ")
print(b)
print("\na + b: ")
print(a+b)
print("\na / b: ")
print(a/b)
print("\na^2: ")
print(a**2)
print("\nmatrix multiplication( a@b.T): ")
print(a @ b.T)

print("\nshape cua a@b.T : ")
print((a@b.T).shape)
```

## Nhận xét:
- phép toán a*b là nhân từng phần tử tương ứng
- phép toán a@b.T là nhân ma trận 
- với a có kích thước 2x3 còn b.T có kích thước 3x2, nên shape của kết quả a@b.T là 2x2


# Bài 7: Các hàm thống kê trên Tensor

```python
a= torch.randn(3,4)
print(a)
print("\nMean: ")
print(torch.mean(a, dim=0))

print("\nStandard deviation: ")
print(torch.std(a, dim=0))

print("\nCumulative sum: ")
print(torch.cumsum(a, dim=0))


print("\nMean: ")
print(torch.mean(a, dim=1))

print("\nStandard deviation: ")
print(torch.std(a, dim=1))

print("\nCumlative sum: ")
print(torch.cumsum(a, dim=1))
```

## Nhận xét:
- dim=0 thực hiện theo chiều đầu tiên nên kết quả thống kê theo từng cột
- dim=1 thực hiện theo chiều thứ 2 nên kết quả thống kê theo từng hàng
- khi đổi dim=0 sang dim=1, phép tính thực hiện theo chiều khác nhau nên kết quả chuyển từ thống kê theo cột sang thống kê theo hàng

# Bài 8: Autograd cơ bản

```python
x= torch.tensor(
    3.6,
    requires_grad=True
)
y= x*x
y.backward()

print("x= ", x)
print("y= ", y)
print("x.grad= ", x.grad)
```
