# Bài 7: Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột. 

```python
a= torch.randn(3,4)
print(a)
print("\nMean: ")
print(torch.mean(a, dim=0))

print("\nStandard deviation: ")
print(torch.std(a, dim=0))

print("\nCumulative sum: ")
print(torch.cumsum(a, dim=0))


print("\nMean: ")
print(torch.mean(a, dim=1))

print("\nStandard deviation: ")
print(torch.std(a, dim=1))

print("\nCumlative sum: ")
print(torch.cumsum(a, dim=1))
```

# Bài 8: Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động. 

```python
x= torch.tensor(
    3.6,
    requires_grad=True
)
y= x*x
y.backward()

print("x= ", x)
print("y= ", y)
print("x.grad= ", x.grad)
```

## kiểm tra gradient
-  ta có: y = x²
- đạo hàm y'= 2x
- tại x=3.6 thì y'= 2 x 3.6 = 7.2
- kết quả bằng với x.grad

# Bài 9: Gradient Descent tìm hệ số đa thức
- Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
polynomial= np.poly1d([1,2,3]) #đa thức thật
N=20 # số lượng dữ liệu

X= np.random.randn(N,1)*5 #sinh dữ liệu X
Y= polynomial(X) #tính Y theo đa thức

XX= np.hstack([  # tạo ma trận đặc trưng [x^2, x, 1]
    X*X,
    X,
    np.ones_like(X)
])

w= torch.randn(3,1, requires_grad=True) #3 hệ số cần tìm

#chuyển dữ liệu sang tensor
x= torch.tensor(XX, dtype=torch.float32)
y= torch.tensor(Y, dtype=torch.float32)

optimizer=torch.optim.NAdam([w], lr=0.01)

print("initial coeficients: ")
print(w)

# huấn luyện
for epoch in range (1000):
    optimizer.zero_grad() # xoá gradient của lần lặp trước
    y_pred= x@w
    mse= torch.mean(torch.square(y-y_pred))
    mse.backward() # tính gradient của hàm mất mát mse đối với các hệ số
    optimizer.step() #dùng gradient để cập nhật các hệ số, làm cho sai số nhỏ hơn

print("learned coeficients: ")
print(w)

print("\nExpected coeficients: ")
print("[1,2,3]")

```

```python
print("\nSo sanh he so:")

print("He so tim duoc:")
print(w.detach().numpy())

print("\nHe so mong doi:")
print("[1, 2, 3]")
```

## So sánh 3 hệ số học được với [1,2,3]
- nhận được 3 hệ số gần với [1,2,3]
## Giải thích vai trò của  optimizer.zero_grad(), mse.backward() và optimizer.step(). 
- optimizer.zero_grad():  xoá gradient của lần lặp trước
- mse.backward(): tính gradient của hàm mất mát mse đối với các hệ số
- optimizer.step(): dùng gradient để cập nhật các hệ số nhằm làm cho sai số nhỏ hơn

# Bài 10: Giải hệ phương trình bằng Autograd 
- Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình. 

```python
import random
A= torch.tensor(random.random(), requires_grad=True)
B= torch.tensor(random.random(), requires_grad=True)
C= torch.tensor(random.random(), requires_grad=True)
D= torch.tensor(random.random(), requires_grad=True)

EPOCHS =2000
optimizer= torch.optim.NAdam([A,B,C,D], lr=0.01)

for _ in range(EPOCHS):
   #sai số của phương trình
   y1= A+B-9
   y2= C-D-1
   y3= A+C-8
   y4= B-D-2

   #tổng bình phương sai số
   sqerr= y1 * y1 + y2 * y2 + y3 * y3 + y4 * y4

   optimizer.zero_grad()
   sqerr.backward()
   optimizer.step()

print("A= ", A)
print("B= ", B)
print("C= ", C)
print("D= ", D)

```

```python
print("\nKiem tra he phuong trinh: ")
print("A + B= ", (A+B).item())
print("C - D= ", (C-D).item())
print("A + C= ", (A+C).item())
print("B - D= ", (B-D).item())
```

## Nhận xét:
- hệ phương trình có nhiều nghiệm vì có 4 ẩn nhưng các phương trình không độc lập để xác định duy nhất cả 4 ẩn. có thể chọn biến tự do, các biến còn lại được xác định theo biến đó
